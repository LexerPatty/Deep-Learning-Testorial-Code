```python
"""
Perceptron :
A perceptron takes some inputs, gives each input a level of importance, adds them together, and makes a decision.

Imagine you are deciding:
"Should I go outside today?"
You might consider:

☀️ Is the weather good?
🌧️ Is it raining?
📅 Is it a holiday?
💼 Do I have work?

You could make a simple rule:

If weather is good AND it is a holiday
        ↓
     GO OUT



STRUCTURE OF PERCEPTRON :

 x₁ ── w₁ ──┐
             │
 x₂ ── w₂ ──┤
             ├──> Σ ──> Activation ──> Output
 x₃ ── w₃ ──┤
             │
 Bias ───────┘

"""
```

```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Let's assume:

# Study Hours → hours studied per day
# Attendance → attendance percentage
# 1 → Pass
# 0 → Fail
```

```python
# z=x1​w1​+x2​w2​+b
```

```python
# Input data
# [Study Hours, Attendance]

X = np.array([
    [2, 5],
    [3, 6],
    [5, 8],
    [6, 7],
])

# Target/output
# 0 = Fail
# 1 = Pass

y = np.array([
    0,
    0,
    1,
    1
])
```

```python
print("Input data:")
print(X)

print("\nActual output:")
print(y)
```

```python
# ==========================================
# 2. Hyperparameters
# ==========================================

weights = np.array([0.0, 0.0])
bias = 0.0

learning_rate = 0.1

epochs = 10
```

```python
# ==========================================
# 3. Prediction Function
# ==========================================

def predict(inputs, weights, bias):

    # Weighted sum:
    # (x1 * w1 + x2 * w2) + bias

    total = np.dot(inputs, weights) + bias

    # Step Activation Function

    if total >= 0:
        return 1
    else:
        return 0
```

```python
# ==========================================
# 4. Training Loop
# ==========================================

for epoch in range(epochs):

    print(f"\n========== Epoch {epoch + 1} ==========")

    for i in range(len(X)):

        # ----------------------------------
        # Step 1: Make Prediction
        # ----------------------------------

        prediction = predict(
            X[i],
            weights,
            bias
        )

        # ----------------------------------
        # Step 2: Calculate Error
        # ----------------------------------

        error = y[i] - prediction

        print(f"\nInput: {X[i]}")
        print(f"Actual: {y[i]}")
        print(f"Prediction: {prediction}")
        print(f"Error: {error}")

        # ----------------------------------
        # Step 3: Update Weights and Bias
        # ----------------------------------

        if error != 0:

            weights += (
                learning_rate
                * error
                * X[i]
            )

            bias += (
                learning_rate
                * error
            )

            print("❌ Wrong prediction → Updating...")

        else:

            print("✅ Correct prediction → No update")

        # ----------------------------------
        # Step 4: Show Updated Parameters
        # ----------------------------------

        print(f"Updated Weights: {weights}")
        print(f"Updated Bias: {bias}")
```

```python
# ==========================================
# 5. Training Complete
# ==========================================

print("\n================================")
print("Training Complete!")
print("================================")

print(f"Final Weights: {weights}")
print(f"Final Bias: {bias}")
```

```python
import numpy as np

# Training data
# 0 = Fail
# 1 = Pass

X = np.array([1, 2, 3, 4, 5, 6])
y = np.array([0, 0, 0, 1, 1, 1])


# Initial values
weight = 0
bias = 0

learning_rate = 0.1


# Training
for epoch in range(10):

    for i in range(len(X)):

        # Weighted sum
        z = X[i] * weight + bias

        # Prediction
        if z >= 0:
            prediction = 1
        else:
            prediction = 0

        # Error
        error = y[i] - prediction

        # Update
        weight = weight + learning_rate * error * X[i]
        bias = bias + learning_rate * error


# Show learned values
print("Learned Weight:", weight)
print("Learned Bias:", bias)


# Test
hours = 5

z = hours * weight + bias

if z >= 0:
    prediction = 1
else:
    prediction = 0


if prediction == 1:
    print("Student will PASS ✅")
else:
    print("Student will FAIL ❌")
```
